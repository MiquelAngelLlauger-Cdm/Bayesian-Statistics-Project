```python
import numpy as np
import matplotlib.pyplot as plt

# -------------------------
# CONFIGURACIÓ
# -------------------------

np.random.seed(42)

K = 6                 # nombre de beacons
sigma_true = 0.4      # soroll real

# mida de l'espai
Lx, Ly = 10, 10

# -------------------------
# POSICIÓ REAL
# -------------------------

x_true = np.array([6.5, 4.0])

# -------------------------
# SENSORS (BEACONS)
# -------------------------

beacons = np.array([
    [0, 0],
    [0, 10],
    [10, 0],
    [10, 10],
    [5, 0],
    [0, 5]
])

# -------------------------
# DISTÀNCIES REALS
# -------------------------

true_distances = np.linalg.norm(
    beacons - x_true,
    axis=1
)

# -------------------------
# MESURES SOROLLOSES
# -------------------------

measurements = (
    true_distances
    + np.random.normal(0, sigma_true, K)
)

# -------------------------
# VISUALITZACIÓ
# -------------------------

plt.figure(figsize=(7,7))

# beacons
plt.scatter(
    beacons[:,0],
    beacons[:,1],
    s=120,
    label='Sensors'
)

# posició real
plt.scatter(
    x_true[0],
    x_true[1],
    c='red',
    s=150,
    label='Posició real'
)

# cercles observats
for i, beacon in enumerate(beacons):
    circle = plt.Circle(
        beacon,
        measurements[i],
        fill=False,
        alpha=0.3
    )
    plt.gca().add_artist(circle)

plt.xlim(-1, 11)
plt.ylim(-1, 11)

plt.grid(True)
plt.legend()
plt.title("Localització amb mesures sorolloses")
plt.show()

# imprimir mesures
print("Mesures:")
for i in range(K):
    print(f"Beacon {i}: {measurements[i]:.3f}")
```
